Numpy array copy vs view

```python
# Copy 
import numpy as np 
arr = np.array([1,2,4,5])
x = arr.copy()
arr[1]=10
print(arr)
print("Copy array not changed",x )
```

```python
# View 
ar = np.array([6,7,8,9])
y = ar.view()
ar[0] = 10
print("Original array:", ar)
print("View array:", y)
```

```python
# Check if array owns its data 
arr_1 = np.array([4,5,7,2])
x = arr_1.copy()
y = arr_1.view()

print(x.base)
print(y.base)
```
