Numpy Data Types 

```python
import numpy as np 
arr = np.array([1,2,3,4])
print(type(arr))
```

```python
# Creating array with defined datatype 
ar = np.array([1,2,3,4], dtype='S')
print(ar)
print(ar.dtype)
```

```python
# Change data type from integer to boolean 
ar_1 = np.array([1,0,3])
new_arr = ar_1.astype(bool)
print(new_arr)
print(type(new_arr))
```
