Create a Numpy ndarray object 

```python
import numpy as np 
arr = np.array([1,2,3,4,5])
print(arr)
print(type(arr))
```

```python
array = np.array([2,3,4,5])
print(type(array))
```

Higher Dimensional Array 

```python
ar = np.array([1,2,3,4], ndmin=5)
print(ar)
```
