Mean Median Mode

Mean: Average Value 
Median: Middle Value 
Mode: Most frequent value 

Mean 

```python
import numpy as np 
speed = [23,30,39,42,50,100,150,250]
x = np.mean(speed)
print(f" Mean of {speed} is :", x)
```

Median 

```python
speed = [23,30,39,42,50,100,150,250]
y = np.median(speed)
print(f"Median of {speed} is :", y)
```

Mode 

```python
from scipy import stats
speed_1 = [12,32,54,34,67,84,54,12,12]
z = stats.mode(speed_1)
print(z)
```
