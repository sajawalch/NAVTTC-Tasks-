```python
# Set Symmetric Difference 
A = {2,3,5}
B = {1,2,6}

# perform difference using & 
print('using ^:', A^B)

# perform difference using symmetric_difference()
print("Using symmetric_difference():", A.symmetric_difference(B))
```
