Numpy Searching Arrays 

```python
import numpy as np 
arr = np.array([1,2,3,4,5,4])
x = np.where(arr==4)
print(x)
```

```python
# find indexes where value are odd 
ar = np.array([10,11,12,14,17])
y = np.where(ar%2==1)
print(y)
```

```python
# Search Sorted 
# find the indexes where the value 7 should be inserted 
ar_1 = np.array([6,7,8,9])
x = np.searchsorted(ar_1, 9)
print(x)
```

```python
arr_2 = np.array([1,2,3,4,5,6,7,8])
z = np.searchsorted(arr_2, [2,7])
print(z)
```
