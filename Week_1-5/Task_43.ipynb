Pandas Series 

```python
import pandas as pd 
a=[1,5,3,2]
var = pd.Series(a)
print(var)
```

```python
print(var[1])
```

Create Labels

```python
b = [5,3,7,9]
myvar = pd.Series(b, index=["a","b","c","d"])
print(myvar)
print(myvar["c"])
```

Key/Value objects as Series 

```python
calories = {"day1":300, "day2":100,"day3":400}
new_var = pd.Series(calories)
print(new_var)
print(type(new_var))
```
