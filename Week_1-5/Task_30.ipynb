Numpy Array Slicing

```python
import numpy as np 
arr = np.array([1,2,4,5])
print(arr[1:3])

# Element from index 2 to end 
print(arr[2:])

# 
print(arr[-2:-4])
```

```python
# Step value 
array = np.array([1,2,3,4,5,6,7])
print(array[1:5:2])

# Every other element from the entire array 
print(arr[::2])
```

```python
# Slicing 2 D array 
ar = np.array([[1,2,3,4], [5,6,7,8]])
print(ar[1, 1:4])
```
