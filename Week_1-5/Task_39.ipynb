Numpy Sorting Arrays

```python
import numpy as np 
arr = np.array([2,1,7,4,9])
print(np.sort(arr))
```

```python
# sort array of string
ar = np.array(['banana', 'apple', 'mango'])
print(np.sort(ar))
```

```python
# Sorting a 2D array 
arr_1 = np.array([[1,6,4], [9,4,8]])
print(np.sort(arr_1))
```
