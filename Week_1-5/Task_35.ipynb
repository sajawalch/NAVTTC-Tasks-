Numpy Array Iteration 

```python
# Iterating 1 D array
import numpy as np 
arr = np.array([1,2,3,4])
for x in arr:
    print(x)
```

```python
# Iterating 2D array 
ar = np.array([[1,2,3,4], [6,7,8,9]])
for x in ar:
    print(x)
```

```python
# Iterate on each scalar element of 2 D array 
arr_1 = np.array([[1,2,3], [4,5,6]])
for x in arr_1:
    for y in x:
        print(y)
```
